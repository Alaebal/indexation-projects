```python
image_gray = cv.cvtColor(img, cv.COLOR_BGR2GRAY )
plt.imshow(image_gray,cmap='gray')
```

```python
image_HSV= cv.cvtColor(img, cv.COLOR_BGR2HSV )
plt.imshow(image_HSV)
```

```python
plt.figure(figsize=(15,20))
plt.subplot(1,3,1)
plt.axis('off')
plt.title("image originale")
plt.imshow(img2)
plt.subplot(1,3,2)
plt.axis('off')
plt.title("image en niveau de gris")
plt.imshow(image_gray, cmap='gray')
plt.subplot(1,3,3)
plt.axis('off')
plt.title("image en HSV")
plt.imshow(image_HSV)
```

```python
plt.figure(figsize=(5,8))
plt.subplot(4,2,1)
plt.axis('off')
plt.title("image en RVB")
plt.imshow(img2)
plt.subplot(4,2,2)
plt.axis('off')
plt.title("image en BVR")
plt.imshow(img)
plt.subplot(4,2,3)
plt.axis('off')
plt.title("image en niveau de gris")
plt.imshow(image_gray, cmap='gray')
plt.subplot(4,2,4)
plt.axis('off')
plt.title("image en HSV")
plt.imshow(image_HSV)
```

```python
plt.figure(figsize=(10,10))
plt.subplot(2,4,1)
plt.axis('off')
plt.title("image en RVB")
plt.imshow(img2)
plt.subplot(2,4,4)
plt.axis('off')
plt.title("image en BVR")
plt.imshow(img)
plt.subplot(2,4,6)
plt.axis('off')
plt.title("image en niveau de gris")
plt.imshow(image_gray, cmap='gray')
plt.subplot(2,4,7)
plt.axis('off')
plt.title("image en HSV")
plt.imshow(image_HSV)
```

```python
print (image_gray.shape)
print (image_HSV.shape)
```

```python
img_egual = cv.equalizeHist(image_gray)
```

```python
res = np.hstack((image_gray,img_egual))
plt.imshow (res, cmap='gray')
```

```python
roihist = cv.calcHist([image_gray],[0], None, [256], [ 0, 256] )
roihist2 = cv.calcHist([img_egual],[0], None, [256], [ 0, 256] )
xs=np.linspace(0,255,256)
plt.subplot(2,1,1)
plt.plot(xs,roihist,color='k')
plt.subplot(2,1,2)
plt.plot(xs,roihist2,color='r')
```

```python
color = ('blue','green','red')
plt.title('histogrammes couleurs de l\'image lena' , fontweight ="bold",fontsize='xx-large')
for i,col in enumerate(color):
 hist = cv.calcHist([img],[i],None,[256],[0,256])
 plt.plot(hist,color = col[0],label=col,linewidth=3.0)
 plt.xlim([0,256])
 legend=plt.legend(loc="upper left")
 legend.set_title("Histogramme")
plt.show()
```

```python
b,g,r=cv.split(img2)
```

```python
hist_b = cv.calcHist([b],[0],None,[256],[0,256])
plt.plot(hist_b,color = 'blue',label="blue",linewidth=3.0)
plt.xlim([0,256])
legend=plt.legend(loc="best")
legend.set_title("Histogramme de couleur blue")
plt.show()
```

```python
hist_g = cv.calcHist([g],[0],None,[256],[0,256])
plt.plot(hist_g,color='g',label ="vert", linewidth=3.0 )
plt.xlim([0,256])
legend=plt.legend(loc="center right")
legend.set_title("Histogramme de couleur vert")
plt.show()
```

```python
hist_r = cv.calcHist([r],[0],None,[256],[0,256])
plt.plot(hist_r,color ='r',label="rouge",linewidth=3.0)
plt.xlim([0,256])
legend=plt.legend(loc="lower right")
legend.set_title("Histogramme de couleur rouge")
plt.show()
```

```python
images=(img_egual,image_gray,b,g,r,)
color = ('yellow','pink','blue','green','red')
labels=('hist_égalisé','hist_gris','hist_blue','hist_green','hist_red')
i=0
plt.figure(figsize=(15,5))
plt.suptitle('Les histogrammes couleurs, niveau de gris et égalisé de l image lena',fontweight='bold',family='cursive',fontsize='large')
for img in (images):
 histr = cv.calcHist([img],[0],None,[256],[0,256])
 plt.plot(histr,color = color[i],label=labels[i],linewidth=1.0)
 i+=1
legend=plt.legend(loc="best")
legend.set_title("hists")
plt.show()
```
